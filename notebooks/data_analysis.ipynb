```python
Q1 = df["P"].quantile(0.25)
Q3 = df["P"].quantile(0.75)

IQR = Q3 - Q1

print("Q1 =", Q1)
print("Q3 =", Q3)
print("IQR =", IQR)
```

```python
filter = (df["P"] >= Q1 - 1.5 * IQR) & (df["P"] <= Q3 + 1.5 * IQR)

df = df.loc[filter]
```

```python
df.shape
```

```python
print("Summer Crops")
print(df[(df["temperature"] > 30) & (df["humidity"] > 50)]["label"].unique())

print("---------------------------------------------")

print("Winter Crops")
print(df[(df["temperature"] < 20) & (df["humidity"] > 30)]["label"].unique())

print("---------------------------------------------")

print("Rainy Crops")
print(df[(df["rainfall"] > 200) & (df["humidity"] > 50)]["label"].unique())
```

```python
y = df["label"]

x = df.drop("label", axis=1)
```

```python
y = df["label"]

x = df.drop("label", axis=1)
```

```python
print("Shape of X:", x.shape)
print("Shape of Y:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=0
)
```

```python
print("X Train:", x_train.shape)
print("X Test :", x_test.shape)
print("Y Train:", y_train.shape)
print("Y Test :", y_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.cluster import KMeans
from sklearn.metrics import accuracy_score
```

```python
model_lr = LogisticRegression(max_iter=500)

model_lr.fit(x_train, y_train)
```

```python
prediction_lr = model_lr.predict(x_test)
```

```python
accuracy = accuracy_score(y_test, prediction_lr)

print("Accuracy:", accuracy)
```

```python
model_knn = KNeighborsClassifier(n_neighbors=5)

model_knn.fit(x_train, y_train)
```

```python
prediction_knn = model_knn.predict(x_test)
```

```python
accuracy_knn = accuracy_score(y_test, prediction_knn)

print("KNN Accuracy:", accuracy_knn)
```

```python
print("Logistic Regression Accuracy:", accuracy)

print("KNN Accuracy:", accuracy_knn)
```

```python
from sklearn.cluster import KMeans
```

```python
wcss = []

for i in range(1,11):
    kmeans = KMeans(n_clusters=i, random_state=0)
    kmeans.fit(x)
    wcss.append(kmeans.inertia_)
```

```python
plt.figure(figsize=(8,5))
plt.plot(range(1,11), wcss, marker='o')
plt.title("Elbow Method")
plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")
plt.show()
```

```python
kmeans = KMeans(n_clusters=4, random_state=0)

kmeans.fit(x)
```
